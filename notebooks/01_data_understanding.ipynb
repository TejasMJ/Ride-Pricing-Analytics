#### Based on the dataset documentation, the meaning of each column in the Dynamic Pricing dataset is as follows:

* Number of Riders - Number of riders requesting rides under the given market conditions.
* Number of Drivers - Number of drivers available to provide rides under the given market conditions.
* Location Category - Category representing the type of location associated with the ride, such as urban, suburban, or rural.
* Customer Loyalty Status - Loyalty category assigned to the customer based on their relationship with the ride-sharing service.
* Number of Past Rides - Number of rides previously taken by the customer.
* Average Ratings - Average rating associated with the customer or their previous ride activity.
* Time of Booking - Time period or category representing when the ride was booked.
* Vehicle Type - Type or category of vehicle selected for the ride.
* Expected Ride Duration - Expected duration of the ride.
* Historical Cost of Ride - Historical fare or cost associated with the ride, serving as the primary target variable for ride-cost prediction.

#### Checking for missing values per column (including empty strings)

```python
# Count of NaN / None values per column
nan_count = df.isnull().sum()

# Count of empty strings per column
empty_count = (df == '').sum()

# Total missing values (NaN + empty strings)
total_missing = nan_count + empty_count

# Percentage of missing values
missing_percentage = (total_missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': total_missing,
    'Percentage (%)': missing_percentage
}, index=df.columns).sort_values(by='Percentage (%)', ascending=False)

print(missing_df)
```
