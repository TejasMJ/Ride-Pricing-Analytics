#### Loading the libraries

```python
# Importing the required libraries

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

#### Loading the dataset

```python
df = pd.read_csv("../data/raw/dynamic_pricing.csv")
```

#### Categorical vs Categorical

Vehicle Type vs Location Category

```python
# Vehicle Type vs Location Category

vehicle_location = pd.crosstab(
    df['Location_Category'],
    df['Vehicle_Type']
)

fig = px.bar(
    vehicle_location,
    barmode='group',
    title='Vehicle Type by Location Category',
    labels={
        'Location_Category': 'Location Category',
        'value': 'Number of Rides',
        'Vehicle_Type': 'Vehicle Type'
    }
)

fig.show()
```
