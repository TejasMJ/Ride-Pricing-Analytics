#### Loading the libraries

```python
# Importing the required libraries

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

#### Loading the dataset

```python
df = pd.read_csv("../data/raw/dynamic_pricing.csv")
```

#### Univariate Analysis

Numerical Variable Distributions

```python
# Numerical columns

numerical_columns = [
    'Number_of_Riders',
    'Number_of_Drivers',
    'Number_of_Past_Rides',
    'Average_Ratings',
    'Expected_Ride_Duration',
    'Historical_Cost_of_Ride'
]
```

Number of Riders

```python
fig = px.histogram(
    df,
    x='Number_of_Riders',
    nbins=20,
    title='Distribution of Number of Riders',
    labels={'Number_of_Riders': 'Number of Riders'}
)

fig.show()
```

Categorical Variable Distributions

Location Category

```python
location_counts = (
    df['Location_Category']
    .value_counts()
    .reset_index()
)

location_counts.columns = ['Location_Category', 'Count']

fig = px.bar(
    location_counts,
    x='Location_Category',
    y='Count',
    title='Distribution of Location Categories',
    labels={
        'Location_Category': 'Location Category',
        'Count': 'Number of Rides'
    }
)

fig.show()
```

Customer Loyalty Status

```python
loyalty_counts = (
    df['Customer_Loyalty_Status']
    .value_counts()
    .reset_index()
)

loyalty_counts.columns = ['Customer_Loyalty_Status', 'Count']

fig = px.bar(
    loyalty_counts,
    x='Customer_Loyalty_Status',
    y='Count',
    title='Distribution of Customer Loyalty Status',
    labels={
        'Customer_Loyalty_Status': 'Customer Loyalty Status',
        'Count': 'Number of Rides'
    }
)

fig.show()
```
